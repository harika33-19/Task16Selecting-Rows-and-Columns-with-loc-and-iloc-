```python
import pandas as pd

data = {
    "Name": ["Ravi", "Priya", "Arjun", "Sneha", "Kiran"],
    "Age": [20, 21, 19, 22, 20],
    "Marks": [85, 92, 78, 88, 95],
    "Department": ["CSE", "ECE", "CSE", "IT", "ECE"]
}

df = pd.DataFrame(data, index=["S1", "S2", "S3", "S4", "S5"])

print(df)
```

```python
print(df.loc["S1"])
```

```python
print(df.loc["S2", "Name"])
```

```python
print(df.loc[["S1", "S3", "S5"]])
```

```python
print(df.loc[:, ["Name", "Marks"]])
```

```python
print(df.loc[["S2", "S4"], ["Name", "Marks"]])
```

```python
print(df.iloc[0])
```

```python
print(df.iloc[1, 0])
```

```python
print(df.iloc[[0, 2, 4]])
```

```python
print(df.iloc[[1, 3], [0, 2]])
```

```python
print(df.iloc[:, [0, 2]])
```

```python
print(df.iloc[[1, 3], [0, 2]])
```
